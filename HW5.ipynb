Оптимизация типов pandas


```python
pd_data = pd.read_csv('Housing.csv')
```

```python
pd_data.describe()
```

Оптимальные типы данных:

price - int32
area - int16
bedrooms - int8
bathrooms - int8
stories - int8
parking - int8


```python
pd_data.dtypes
```

```python
pd_data.head()
```

```python
pd_data.info()
```

```python
pd_data['price'] = pd_data['price'].astype('int32')
pd_data['area'] = pd_data['area'].astype('int16')
pd_data['bedrooms'] = pd_data['bedrooms'].astype('int8')
pd_data['bathrooms'] = pd_data['bathrooms'].astype('int8')
pd_data['stories'] = pd_data['stories'].astype('int8')
pd_data['parking'] = pd_data['parking'].astype('int8')
```
