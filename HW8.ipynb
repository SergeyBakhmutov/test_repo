```python
api_base_url = "https://www.anapioficeandfire.com/api"
all_books_list = []

count = 0
page = 1

while True:
    response = requests.get(f'{api_base_url}/books?page={page}')
    data = response.json()

    if not data:
        break

    all_books_list.extend(data)
    count += 1
    page += 1
```

```python
pd_with_books = pd.DataFrame(all_books_list)
```

```python
api_base_url = "https://www.anapioficeandfire.com/api"
all_houses_list = []

count = 0
page = 1

while True:
    response = requests.get(f'{api_base_url}/houses?page={page}')
    data = response.json()

    if not data:
        break

    all_houses_list.extend(data)
    count += 1
    page += 1
```
